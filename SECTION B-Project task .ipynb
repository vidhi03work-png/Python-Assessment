```python
import pandas as pd
```

```python
#1
df = pd.read_csv("sales_data_sample.csv",encoding="latin")
print(df.head())

df["PRODUCTLINE"] = df["PRODUCTLINE"].str.strip().str.title()

print(df["PRODUCTLINE"].head(10))

print(df["PRODUCTLINE"].unique())
```

```python
#2
import re
cleaned_sales = []

for value in df["SALES"]:

    value = str(value)

    value = re.sub(r"[^0-9.]", "", value)

    value = float(value)

    cleaned_sales.append(value)

df["SALES"] = cleaned_sales

print("QUESTION 2: CLEANED SALES")
print(df["SALES"].head(10))

print("SALES Data Type:", df["SALES"].dtype)
```

```python
#3

product_sales = {}

for product, sales in zip(df["PRODUCTLINE"], df["SALES"]):

    if product in product_sales:
        product_sales[product] += sales

    else:
        product_sales[product] = sales

print("Total Sales by Product Line:")

for product, total in product_sales.items():
    print(product, ":", round(total, 2))
```

```python
#4

for product, total in product_sales.items():

    if total > 100000:
        category = "High Revenue Product"
    else:
        category = "Not a High Revenue Product"

    print(product, ":", category)
```
